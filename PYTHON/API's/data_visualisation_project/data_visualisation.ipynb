```python
import pandas as pd
import matplotlib.pyplot as plt

# Monthly revenue for 3 years
dates = pd.date_range("2023-01-01", periods=36, freq="MS")

revenue = [
    12000, 13500, 12800, 14500, 15000, 16200,
    15800, 17000, 18000, 17500, 19500, 22000,
    20000, 21500, 21000, 23000, 24500, 25000,
    24000, 26000, 27500, 26800, 29000, 32000,
    30000, 31500, 31000, 33500, 35000, 36000,
    34500, 37000, 39000, 38000, 41000, 45000
]

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(dates, revenue, marker="o")

ax.set_title("Monthly Revenue Shows an Overall Upward Trend")
ax.set_xlabel("Month")
ax.set_ylabel("Revenue (₹)")

fig.autofmt_xdate()
fig.tight_layout()

fig.savefig(
    "charts/01_monthly_revenue.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)

data = pd.DataFrame(
    np.random.randn(100, 10),
    columns=[
        "Sales", "Profit", "Revenue", "Costs", "Ads",
        "Customers", "Orders", "Discount", "Returns", "Stock"
    ]
)

# Calculate correlations between numeric columns
correlation = data.corr()

fig, ax = plt.subplots(figsize=(10, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    ax=ax
)

ax.set_title("Ten Numeric Variables Show Different Correlation Strengths")
ax.set_xlabel("Numeric Variables")
ax.set_ylabel("Numeric Variables")

fig.savefig(
    "charts/06_correlation_heatmap.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

categories = [
    "Electronics", "Clothing", "Furniture", "Books",
    "Beauty", "Sports", "Toys", "Groceries"
]

revenue = [95000, 72000, 88000, 45000, 61000, 54000, 39000, 80000]

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Honest chart: bar axis starts at zero
axes[0].bar(categories, revenue)
axes[0].set_ylim(0, 105000)
axes[0].set_title("Honest View: Revenue Differences in Context")
axes[0].set_xlabel("Product Category")
axes[0].set_ylabel("Revenue (₹)")
axes[0].tick_params(axis="x", rotation=45)

# Misleading chart: truncated y-axis exaggerates differences
axes[1].bar(categories, revenue)
axes[1].set_ylim(30000, 100000)
axes[1].set_title("Misleading View: Differences Look Much Larger")
axes[1].set_xlabel("Product Category")
axes[1].set_ylabel("Revenue (₹)")
axes[1].tick_params(axis="x", rotation=45)

fig.tight_layout()

fig.savefig(
    "charts/honest_vs_truncated.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()
```
